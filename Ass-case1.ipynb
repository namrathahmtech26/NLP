```python
import nltk
import re

nltk.download("punkt")
nltk.download("punkt_tab")
```

```python
reviews = [
    "The product quality is excellent! I am very satisfied with the purchase. "
    "Will definitely recommend it.",
    "Terrible customer service. I had to wait two weeks for a response. "
    "Not buying again!",
    "Average product. Delivery was on time, but the packaging was damaged."
]
```

```python
positive_words = {"excellent", "satisfied", "recommend", "good", "great", "love"}
negative_words = {"terrible", "damaged", "bad", "poor", "disappointed", "hate"}
```

```python
tokenized_reviews = []

for review_number, review in enumerate(reviews, start=1):
    sentences = nltk.sent_tokenize(review)
    tokenized_reviews.append(sentences)

    print(f"\nReview {review_number}:")
    print(f"Sentence count: {len(sentences)}")

    found_positive = set()
    found_negative = set()

    for sentence_number, sentence in enumerate(sentences, start=1):
        tokens = nltk.word_tokenize(sentence)
        # Exclude punctuation when counting words.
        words = [token for token in tokens if re.search(r"[A-Za-z0-9]", token)]

        print(f"  Sentence {sentence_number}: {sentence}")
        print(f"    Word tokens: {tokens}")
        print(f"    Word count: {len(words)}")

        for word in words:
            normalized = word.casefold()
            if normalized in positive_words:
                found_positive.add(normalized)
            if normalized in negative_words:
                found_negative.add(normalized)

    print(f"Positive words found: {sorted(found_positive)}")
    print(f"Negative words found: {sorted(found_negative)}")
```

```python
def flag_long_reviews(review_sentences, limit=3):
    """Return True for reviews with more than `limit` sentences."""
    return [len(sentences) > limit for sentences in review_sentences]


flags = flag_long_reviews(tokenized_reviews)

print("Reviews flagged for further analysis:")
for review_number, flagged in enumerate(flags, start=1):
    if flagged:
        print(f"Review {review_number}")
```
